# 01 — Explore the cached CFBD data

Loads every dataset from `data/raw/` and answers the questions that
shape the transfer dataset build in `src/build_dataset.py`. Findings are summarized
at the bottom.

```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
from build_dataset import load, normalize_name  # reuse the pipeline's loader

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 200)

NAMES = ["portal", "ppa", "usage", "stats", "recruiting", "teams_fbs"]
data = {name: load(name) for name in NAMES}
portal, ppa, usage, stats, rec, teams = (data[n] for n in NAMES)
```

## Overview of every dataset: shape, dtypes, missing values, sample rows

```python
def overview(name, df, n=3):
    print(f"===== {name}: {df.shape[0]:,} rows x {df.shape[1]} cols "
          f"(years {df.pull_year.min()}-{df.pull_year.max()}) =====")
    summary = pd.DataFrame({
        "dtype": df.dtypes.astype(str),
        "missing": df.isna().sum(),
        "missing_pct": (df.isna().mean() * 100).round(1),
    })
    display(summary)
    display(df.head(n))

for name in NAMES:
    cols = [c for c in data[name].columns if not c.startswith("logos")]
    overview(name, data[name][cols])
```

## What does the portal `season` field mean?

Two checks: (1) when do transfer dates fall relative to `season`, and (2) which team
do well-known transfers appear with in the production data for that season.

```python
print("season always equals the file's pull year:", (portal.season == portal.pull_year).all())
months = portal.assign(month=portal.transferDate.str[:7])
display(months.groupby("season").month.agg(["min", "max"]))
```

```python
KNOWN = [("Caleb", "Downs"), ("Cameron", "Ward"), ("Cam", "Ward"), ("Michael", "Penix Jr."),
         ("Bo", "Nix"), ("Jayden", "Daniels"), ("Will", "Howard")]
prod = stats[["season", "playerId", "player", "team"]].drop_duplicates()
for first, last in KNOWN:
    p = portal[(portal.firstName == first) & (portal.lastName == last)]
    if p.empty:
        continue
    print(f"\n## {first} {last}")
    display(p[["season", "origin", "destination", "transferDate"]])
    display(prod[prod.player.map(normalize_name).str.endswith(normalize_name(last))
                 & prod.player.str.startswith(first[:3])].sort_values("season"))
```

**Result:** `season` is the **first season the player plays at the new school**. Transfer
dates for season *S* run from fall *S-1* through summer *S*. Caleb Downs (2024) shows up
for Alabama in 2023 and Ohio State in 2024. Cam Ward (2024; listed as "Cameron Ward" in
stats) shows up for Washington State in 2023 and Miami in 2024. So BEFORE = origin in
*S-1* and AFTER = destination in *S*.

## Missing destinations and repeated players

```python
nodest = portal.destination.isna()
print(f"No destination: {nodest.sum():,} of {len(portal):,} ({nodest.mean():.1%})")
display(portal.assign(no_dest=nodest).groupby("season").no_dest.agg(["sum", "mean"]))
display(pd.crosstab(portal.eligibility.fillna("NA"), nodest.rename("no_destination")))
print("destination == origin (returned to same school):", (portal.destination == portal.origin).sum())
```

```python
portal["name_norm"] = (portal.firstName + " " + portal.lastName).map(normalize_name)
seasons_per_name = portal.groupby("name_norm").season.nunique()
print("names in the portal in more than one season:", (seasons_per_name > 1).sum())

same_cycle = portal[portal.duplicated(["season", "name_norm", "origin"], keep=False)]
print("rows sharing season + name + origin (same player listed twice in one cycle):", len(same_cycle))
display(same_cycle.sort_values(["season", "name_norm"])[
    ["season", "firstName", "lastName", "position", "origin", "destination", "transferDate", "stars"]])

print("rows sharing season + name with DIFFERENT origins (mostly common names):",
      portal.duplicated(["season", "name_norm"], keep=False).sum()
      - portal.duplicated(["season", "name_norm", "origin"], keep=False).sum())

# Example of a genuine multi-transfer player.
display(portal[portal.name_norm == "dillon gabriel"][["season", "origin", "destination"]])
```

## PPA and usage structure

```python
print("PPA columns:", [c for c in ppa.columns])
print("Usage columns:", [c for c in usage.columns])
print("\nPPA positions:", ppa.position.value_counts().to_dict())
```

```python
dups = ppa[ppa.duplicated(["season", "id"], keep=False)]
differing = [c for c in ppa.columns
             if dups.groupby(["season", "id"])[c].nunique(dropna=False).max() > 1]
print(f"PPA rows sharing season+id: {len(dups):,}; columns that differ within them: {differing}")
display(dups.sort_values(["season", "id"])[["season", "id", "name", "team", "conference", "totalPPA.all"]].head(4))
print("usage rows sharing season+id:", usage.duplicated(["season", "id"], keep=False).sum())

value_cols = [c for c in ppa.columns if c.startswith(("totalPPA", "averagePPA"))]
n_vals = dups.groupby(["season", "id", "team"])["totalPPA.all"].nunique()
print(f"blank-conference repeats: {(n_vals == 1).sum()} | split across positions: {(n_vals > 1).sum()}")
display(dups[(dups.id == "4677280") & (dups.season == 2022)][["season", "name", "team", "position", "totalPPA.all", "averagePPA.all"]])
```

PPA is nested JSON: `averagePPA` and `totalPPA` each split into all / pass / rush /
firstDown / secondDown / thirdDown / standardDowns / passingDowns, which `json_normalize`
flattens to `averagePPA.all`, `totalPPA.pass`, etc. Usage has the same split under
`usage.*` (with `overall` instead of `all`). **Both cover only offensive skill players**
(QB/RB/WR/TE/FB).

The 353 duplicated player-team-seasons come in two kinds: 283 are exact repeats
except for a blank `conference` (drop one), and **70 are players CFBD split across
two positions** (e.g. RB and WR) with their PPA divided between the rows. Those
have to be **combined**: sum the totals and recompute averages from implied plays.

## Stats: long or wide?

```python
print("rows per player-team-season:")
display(stats.groupby(["season", "playerId", "team"]).size().describe())
display(stats[stats.player == "Caleb Downs"].head(10))
display(stats.groupby("category").statType.unique().to_frame())
print("stat column dtype:", stats.stat.dtype, "| sample values:", stats.stat.unique()[:8])
print("players with >1 team in a season:",
      (stats.groupby(["season", "playerId"]).team.nunique() > 1).sum())
```

Stats are **long format**: one row per player × team × season × (category, statType),
with `stat` stored as a **string**. It has to be converted to numbers and pivoted wide.
It covers all positions, but only players who recorded at least one stat.

## ID fields and links between datasets

```python
id_fields = {
    "portal": "(none)", "ppa": "id", "usage": "id", "stats": "playerId",
    "recruiting": "id (recruit), athleteId", "teams_fbs": "id (team)",
}
display(pd.Series(id_fields, name="id fields"))

key = lambda df, col: set(zip(df.season, df[col].astype(str)))
print("ppa (season,id) found in usage:", f"{len(key(ppa,'id') & key(usage,'id')) / len(key(ppa,'id')):.1%}")
stat_ids = set(stats.playerId.astype(str))
print("ppa id found in stats playerId:", f"{ppa.id.astype(str).isin(stat_ids).mean():.1%}")
print("recruiting athleteId found in stats playerId:", f"{rec.athleteId.astype(str).isin(stat_ids).mean():.1%}")
```

```python
# Does an athlete's ID survive a transfer?
for name in ["Cameron Ward", "Caleb Downs", "Bo Nix"]:
    display(stats[stats.player == name][["season", "playerId", "player", "team"]].drop_duplicates())
```

PPA `id` = usage `id` = stats `playerId` (the CFBD athlete ID). It usually stays the
same through a transfer (Cam Ward, Caleb Downs), but **not always**: Bo Nix has
different IDs at Auburn and Oregon. The portal has **no player ID**, so portal rows have
to be matched by name + team + season. Recruiting `athleteId` links to stats for ~43% of
recruits (many never record a stat).

## Team name spelling across datasets

```python
fbs_all = set(teams.school)
display(teams.groupby("pull_year").school.nunique().rename("FBS teams"))
for label, s in [("portal origin", portal.origin), ("portal destination", portal.destination),
                 ("ppa", ppa.team), ("usage", usage.team), ("stats", stats.team),
                 ("recruiting committedTo", rec.committedTo)]:
    names = set(s.dropna())
    print(f"{label:24} {len(names):4} names | FBS names used: {len(names & fbs_all):3} "
          f"| FBS names never used: {sorted(fbs_all - names)[:5]}")
```

```python
import difflib
# Non-FBS names that look like an FBS spelling variant
all_names = set(portal.origin) | set(portal.destination.dropna()) | set(stats.team) | set(ppa.team)
near = {n: difflib.get_close_matches(n, fbs_all, 1, 0.85) for n in all_names - fbs_all}
display({k: v for k, v in near.items() if v})
display(teams.drop_duplicates("school")[teams.drop_duplicates("school").school.isin(
    ["Miami", "Miami (OH)", "UConn", "Hawai'i", "San José State", "App State", "UL Monroe"])][
    ["school", "alternateNames"]])
print("conference labels -- ppa:", sorted(ppa.conference.dropna().unique())[:8],
      "| stats:", sorted(stats.conference.dropna().unique())[:8])
```

Team names are **consistent**: every FBS school name in `teams_fbs` appears verbatim in
portal, PPA, usage and stats, and the only near-misses (Bowie State vs Boise State, etc.)
are different schools. Non-FBS names are FCS/D2 programs. Conference labels are *not*
consistent (PPA uses `B1G`, stats uses `Big Ten`), but they aren't used for joining.

## Summary of findings

| Question | Finding | Consequence for the build |
|---|---|---|
| Portal `season` | First season at the **new** school (Downs: Alabama 2023 → Ohio State 2024; Ward: WSU 2023 → Miami 2024) | BEFORE = origin, season−1; AFTER = destination, season |
| No destination | 3,971 of 14,422 (27.5%); includes all 256 "Withdrawn" rows | Dropped by the FBS-destination filter |
| destination == origin | 142 rows (player returned to the same school) | Dropped as a separate, logged step (configurable) |
| Repeat players | ~1,900 normalized names appear in more than one portal season (real repeat transfers like Dillon Gabriel 2022 and 2024) | Each portal entry is its own transfer row |
| Same-cycle duplicates | 7 pairs share season + name + origin; most have one null destination | Exact repeats deduped (keep latest) |
| PPA / usage | Nested `averagePPA` / `totalPPA` / `usage` dicts flattened to `x.all`, `x.pass`, `x.rush`, down splits. **Offense only** (QB/RB/WR/TE/FB). 353 duplicated keys: 283 blank-conference repeats, 70 players split across two positions | Repeats dropped, splits combined; defenders/OL get no PPA |
| Stats | **Long** format, `stat` stored as a string, all positions, only players who recorded a stat | Converted to numbers, key stats pivoted wide |
| IDs | Portal has none. PPA `id` = usage `id` = stats `playerId`. Usually stable across a transfer but not always (Bo Nix). Recruiting `athleteId` links to stats for ~43% | Match by name; use ID only to link the other side |
| Team names | Consistent across all datasets for FBS schools (Army and Navy never appear as a portal destination); conference labels differ | Join directly on team name |

**Survivorship caveat:** production data only includes players with at least one
recorded stat. OL, long snappers and players who never got on the field will have
no AFTER match even when the name is right. That is real signal, not a matching failure.
